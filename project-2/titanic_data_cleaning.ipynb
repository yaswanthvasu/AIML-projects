```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("//content/Titanic-Dataset.csv")
```

```python
df.shape
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df = df.drop(columns=['Cabin'])
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3 - Q1
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
```

```python
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[(df['Age'] < lower_limit) | (df['Age'] > upper_limit)]

outliers
```

```python
df['Age'] = np.where(df['Age'] < lower_limit, lower_limit, df['Age'])

df['Age'] = np.where(df['Age'] > upper_limit, upper_limit, df['Age'])
```

```python
outliers = df[(df['Age'] < lower_limit) | (df['Age'] > upper_limit)]
len(outliers)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("//content/Titanic-Dataset.csv")
```

```python
df.sort_values(by='Fare', ascending=False)
```

```python
df.groupby('Sex')['Age'].mean()
```

```python
df.groupby('Sex')['Survived'].sum()
```

```python
df['Survived'].mean()
```

```python
df.groupby('Sex')['Survived'].mean()
```

```python
plt.bar(df['Sex'], df['Survived'])
plt.xlabel('Gender')
plt.ylabel('Survived')
plt.title('Survival count by gender')
plt.show()
```

```python
plt.plot(df['Pclass'], df['Fare'])
plt.xlabel('Passenger Class')
plt.ylabel('Average Fare')
plt.title('Fare by Class')
plt.show()
```

```python
plt.hist(df['Age'], bins=10)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution')
plt.show()
#
```

```python
plt.scatter(df['Age'], df['Fare'])
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Age vs Fare')
plt.show()
#
```
