```python
df=df.drop_duplicates()
```

```python
df['chlorides']=df.chlorides.fillna(df['chlorides'].median())
```

```python

columns = ['fixed acidity', 'volatile acidity', 'citric acid',
           'residual sugar', 'pH', 'sulphates']

for col in columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
Q1=df.quality.quantile(0.25)
Q3=df.quality.quantile(0.75)
IQR=Q3-Q1
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
```

```python
outliers = df[(df['quality'] < lower_limit) | (df['quality'] > upper_limit)]

len(outliers)
```

```python
df['alcohol'] = np.where(df['alcohol'] < lower_limit, lower_limit, df['alcohol'])

df['alcohol'] = np.where(df['alcohol'] > upper_limit, upper_limit, df['alcohol'])
```

```python
df.sort_values(by='fixed acidity',ascending=False)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['alcohol'], df['quality'])
plt.xlabel('Type')
plt.ylabel('Alcohol')
plt.title('Type vs Alcohol')
plt.show()
```

```python
df.groupby('quality')['pH'].mean().plot(kind='line', marker='o')

plt.title("Average pH by Quality")
plt.xlabel("Quality")
plt.ylabel("Average pH")
plt.show()
```

```python
plt.bar(df['type'],df['alcohol'])
plt.xlabel('type')
plt.ylabel('Alcohol')
plt.title('type vs Alcohol')
plt.show()
```

```python
df['alcohol'].plot(kind='hist', bins=10)

plt.title("Alcohol Distribution")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.show()

```
