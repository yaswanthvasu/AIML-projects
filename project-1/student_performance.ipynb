```python
df.select_dtypes(include='number')
```

```python
numeric_df = df.select_dtypes(include='number')

Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)
IQR = Q3 - Q1

outliers = ((numeric_df < (Q1 - 1.5 * IQR)) | (numeric_df > (Q3 + 1.5 * IQR)))

print(outliers.sum())
```

```python
numeric_df = numeric_df.clip(lower=Q1 - 1.5 * IQR,
                             upper=Q3 + 1.5 * IQR,
                             axis=1)

df[numeric_df.columns] = numeric_df
```

```python
df.groupby('sex')['G3'].mean().plot(kind='bar')

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")
plt.show()
```

```python
df.groupby('age')['G3'].mean().plot(kind='line')

plt.title("Average Final Grade by Age")
plt.xlabel("Age")
plt.ylabel("Average Final Grade")
plt.grid(True)
plt.show()
```

```python
plt.hist(df['G3'])

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(df["G1"], df["G3"])

plt.title("G1 vs G3")
plt.xlabel("First Period Grade")
plt.ylabel("Final Grade")
plt.show()
```
